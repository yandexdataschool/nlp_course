## Text generation

**Comparison of strategies for language model text generation:**

| Strategy | Description | Pros & Cons | When to Use |
| --- | --- | --- | --- |
| Greedy Search | Chooses the word with the highest probability as the next word. | **Pros:** Simple and fast. <br> **Cons:** Can lead to repetitive and incoherent text. | Deterministic tasks, debugging |
| Sampling with Temperature | Introduces randomness in word selection. Higher temperature = more randomness. | **Pros:** Allows exploration and diverse output. <br> **Cons:** Too high can lead to nonsensical outputs. | Creative tasks (temperature 0.7-1.0) |
| Nucleus Sampling (Top-p) | Selects from the smallest set of words whose cumulative probability exceeds p. | **Pros:** Balances diversity and quality well. <br> **Cons:** Setting optimal 'p' can be tricky. | **Most common choice** (p=0.9) |
| Beam Search | Explores multiple hypotheses at each step, keeps 'k' most likely. | **Pros:** More reliable than greedy. <br> **Cons:** Can lack diversity. | Machine translation, summarization |
| Top-k Sampling | Randomly selects from top 'k' words with highest probabilities. | **Pros:** Simple diversity control. <br> **Cons:** Fixed k doesn't adapt to distribution. | When you know the desired diversity level |
| Min-p Sampling (New!) | Filters tokens below a minimum probability threshold. | **Pros:** Adaptive to context. <br> **Cons:** Still relatively new. | Alternative to top-p |

**Modern best practices (2024-2025):**
- For creative tasks: `do_sample=True, top_p=0.9, temperature=0.8`
- For factual tasks: `do_sample=True, top_p=0.95, temperature=0.7` or greedy
- For diverse outputs: `do_sample=True, top_p=0.9, temperature=1.0, top_k=50`

#### Low-level code for text generation

````python
prompt = "Moscow is the capital of"
# prompt = "Skippy, a young android, likes to dream about electric"

print(prompt, '\n')

voc = tokenizer.get_vocab()
voc_rev = {v:k for k, v in voc.items()}  # reverse vocab for decode

for i in range(10):
    inputs = tokenizer(prompt, return_tensors='pt', return_token_type_ids=False).to(device)
    logits = model.forward(**inputs).logits[0, -1, :]
    probs = torch.nn.functional.softmax(logits, dim=-1)
    next_token_id = torch.multinomial(probs.flatten(), num_samples=1)

    next_token = tokenizer.decode(next_token_id)
    prompt += next_token

    sorted_probs, sorted_indices = torch.sort(probs, descending=True)
    top_tokens = sorted_indices[:5]
    print(f"Step #{i} candidates:")
    for t, p in zip (top_tokens, sorted_probs):
        t = voc_rev[t.item()]
        print(f"{t:<10}: {p:.4f} ")

    print(f'\nChosen token: {next_token}', end='\n\n', flush=True)
````

The tricky part is when to stop generating: if you don't control for this, your model can accidentally generate a whole new question - and promptly answer it :) To make sure you get the correct answer, stop generating tokens when the model is done explaining its solution. To circumvent this, you need to __stop generating as soon as the model generates "Correct Answer: [A-E]"__

**Stopping Criteria Options:**

1. **Simple approach**: Generate enough tokens (e.g., `max_new_tokens=200`) and parse the output afterwards
2. **Manual generation**: Use the low-level generation code above and stop when you detect the pattern
3. **StoppingCriteria class**: Use transformers' built-in mechanism (example below)

```python
from transformers import StoppingCriteria, StoppingCriteriaList

class AnswerStoppingCriteria(StoppingCriteria):
    def __init__(self, tokenizer, stop_string="Correct Answer:"):
        self.tokenizer = tokenizer
        self.stop_string = stop_string
        
    def __call__(self, input_ids, scores, **kwargs):
        # Decode recent tokens and check for stop string
        decoded = self.tokenizer.decode(input_ids[0][-20:])
        return self.stop_string in decoded

stopping = StoppingCriteriaList([AnswerStoppingCriteria(tokenizer)])
output = model.generate(..., stopping_criteria=stopping)
```

The **self-consistency** technique ([Wang et al., 2023](https://arxiv.org/abs/2203.11171)) works as follows: first, you generate k (e.g. 20-50) "attempts" at an answer using nucleus sampling (or a similar technique with temperature > 0).
Then, you count how many of those attempts chose a particular option (A, B, etc) as the final answer. The option that was chosen most frequently has the most "votes", and therefore "wins".

This is particularly effective for chain-of-thought prompting because the model can explore different reasoning paths, and the correct answer often emerges as the most common conclusion.

To speed up voting, you may want to generate these attempts in parallel as a batch. That should be very easy to implement: just run `model.generate` on a batch with multiple copies of the same prompt.
